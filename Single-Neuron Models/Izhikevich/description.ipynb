```python
import numpy as np
import matplotlib.pyplot as plt
```

# Model

```python
def Model(a, b, c, d, dt: float, T: float):
    N = int(T / dt)

    v = c
    u = c * b

    I = 5.0 * np.ones(N)

    
    v_rec = np.zeros(N)
    t_rec = np.arange(0, T, dt)

    for t in range(N):
        if v >= 30:
            v = c
            u = u + d

        v = (0.04 * np.pow(v, 2) + 5.0 * v + 140.0 - u + I[t]) * dt + v
        u = (a * (b * v - u)) * dt + u
        
        v_rec[t] = v


    return v_rec, t_rec
    
```

# Regular Spiking (RS)

```python
V, T = Model(0.02, 0.2, -65.0, 8.0, 0.01, 300)
plt.plot(T, V)
plt.show()
```

# Intrinsically Bursting (IB)

```python
V, T = Model(0.02, 0.2, -55.0, 4.0, 0.1, 200)
plt.plot(T, V)
plt.show()
```

# Chattering (CH)

```python
V, T = Model(0.02, 0.2, -50.0, 2.0, 0.1, 200)
plt.plot(T, V)
plt.show()
```

# Fast Spiking (FS)

```python
V, T = Model(0.03, 0.2, -65.0, 2.0, 0.1, 200)
plt.plot(T, V)
plt.show()
```
